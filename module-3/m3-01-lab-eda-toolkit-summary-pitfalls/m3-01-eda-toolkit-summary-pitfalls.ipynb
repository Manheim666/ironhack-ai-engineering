```python
import pandas as pd
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
import scipy as sp

np.random.seed(42)
sns.set_style('whitegrid')
```

```python
dataset_A = np.random.normal(loc=50, scale=10, size=500)
dataset_A = pd.DataFrame(dataset_A, columns=['Value'])
dataset_B = np.random.exponential(scale=50, size=500)
dataset_B = pd.DataFrame(dataset_B, columns=['Value'])
dataset_C_1 = np.random.normal(loc=30, scale=50, size=250)
dataset_C_2 = np.random.normal(loc=70, scale=50, size=250)
dataset_C = np.concatenate([dataset_C_1, dataset_C_2])
dataset_C = pd.DataFrame(dataset_C, columns=['Value'])
```

```python
dataset_A
```

```python
print('dataset_A:')
print('Mean:',dataset_A['Value'].mean())
print('Median:',np.median(dataset_A['Value']))
print('Mode',float(sp.stats.mode(dataset_A['Value'])[0]))
print('std:',dataset_A['Value'].std())
print('IQR:',sp.stats.iqr(dataset_A['Value']))
print('Skew:',sp.stats.skew(dataset_A['Value']))
print('Kurtosis:',sp.stats.kurtosis(dataset_A['Value']))
```

```python
print('dataset_B:')
print('Mean:',dataset_B['Value'].mean())
print('Median:',np.median(dataset_B['Value']))
print('Mode',float(sp.stats.mode(dataset_B['Value'])[0]))
print('std:',dataset_B['Value'].std())
print('IQR:',sp.stats.iqr(dataset_B['Value']))
print('Skew:',sp.stats.skew(dataset_B['Value']))
print('Kurtosis:',sp.stats.kurtosis(dataset_B['Value']))
```

```python
print('dataset_C:')
print('Mean:',dataset_C['Value'].mean())
print('Median:',np.median(dataset_C['Value']))
print('Mode',float(sp.stats.mode(dataset_C['Value'])[0]))
print('std:',dataset_C['Value'].std())
print('IQR:',sp.stats.iqr(dataset_C['Value']))
print('Skew:',sp.stats.skew(dataset_C['Value']))
print('Kurtosis:',sp.stats.kurtosis(dataset_C['Value']))
```

```python
datasets = [dataset_A, dataset_B, dataset_C]
dataset_names = ['Dataset A: Symmetric', 'Dataset B: Right-skewed', 'Dataset C: Bimodal']

plt.figure(figsize=(18, 12))

for i in range(3):
    data = datasets[i]['Value']
    mean_val = np.mean(data)
    median_val = np.median(data)
    
    plt.subplot(3, 3, i*3 + 1)
    sns.histplot(data, bins=30, kde=True, color='skyblue')
    plt.title(f'{dataset_names[i]} - Histogram + KDE')
    plt.xlabel('Value')
    plt.ylabel('Frequency')
    
    plt.subplot(3, 3, i*3 + 2)
    sns.boxplot(x=data, orient='h', color='lightgreen')
    plt.title(f'{dataset_names[i]} - Boxplot')
    plt.xlabel('Value')
    
    plt.subplot(3, 3, i*3 + 3)
    sns.kdeplot(data, fill=True, color='salmon')
    plt.axvline(mean_val, color='red', linestyle='--', label='Mean')
    plt.axvline(median_val, color='blue', linestyle='-', label='Median')
    plt.title(f'{dataset_names[i]} - KDE with Mean & Median')
    plt.xlabel('Value')
    plt.ylabel('Density')
    plt.legend()

plt.tight_layout()
plt.show()
```

### Outlier Stress Test — Interpretation

After replacing the five largest values in Dataset A with extreme outliers (by multiplying them by 10), several summary statistics changed noticeably. The statistics that changed the most were the **mean, variance, and standard deviation**. These measures are highly sensitive to extreme values because they depend on the magnitude of all observations in the dataset. When the largest values were increased dramatically, the mean shifted upward and the measures of spread (variance and standard deviation) increased substantially.

On the other hand, the **median and the interquartile range (IQR)** remained relatively stable. These statistics are considered **robust** because they are based on the position of values in the ordered dataset rather than their exact size. The median only reflects the middle observation, so a few extremely large values at the top of the distribution do not significantly affect it. Similarly, the IQR measures the spread of the middle 50% of the data and therefore excludes the extreme outliers.

If I were reporting a **“typical” value** for the modified dataset, I would choose the **median rather than the mean**. The median better represents the central tendency of the majority of observations because it is resistant to extreme values. In contrast, the mean becomes misleading in the presence of large outliers since it is pulled upward by those extreme observations. Therefore, the median provides a more accurate representation of the typical value in this modified dataset.
