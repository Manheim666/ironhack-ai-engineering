```python
import re
import pandas as pd

def validate_registrations(entries):
    valid = []
    invalid = []
    
    email_pattern = r'^[^@]+@[^@]+\.[^@]+$'
    
    for i in entries:
        if i['username'] == '':
            invalid.append(i)
            continue
    
        if not re.match(email_pattern, i['email']):
            invalid.append(i)
            continue
        
        try:
            i['age'] = abs(int(i['age']))
        except (TypeError, ValueError):
            invalid.append(i)
            continue

        valid.append(i)

    return valid, invalid

valid, invalid = validate_registrations(dataset_A)

print(valid)

print(invalid)
```

Type preservation
Parquet retained the original dtypes most faithfully (numeric columns stayed as float/int64), CSV converted everything to object/float64 and JSON had similar issues with timestamps stored as object strings.  
Size- Parquet was the smallest file by a large margin (it’s a binary columnar format with compression). CSV was second, and JSON the largest because of verbose text and repetition.
  
When to choose each:  
  CSV is portable and human‑readable—good for quick exports or systems without special readers.  
  JSON works well for streaming records or web APIs, but can be bulky and loses type fidelity.  
  *Parquet* shines for analytics pipelines where you need compact storage and accurate types; ideal for big data and when using pandas, Spark, or similar.

Parquet is columnar: each column is stored separately. When you read only col_5, pandas reads just that column from disk — no I/O for other columns.

CSV is row-based: each row contains all columns. To read col_5, pandas must parse every row, so selective reads are not possible, making CSV slower for partial column access

```python
import pandas as pd
import numpy as np
from faker import Faker
import random

fake = Faker()
np.random.seed(42)
random.seed(42)

formats = ["Paperback", "E-book"]

flat_rows = []

for _ in range(12):
    title = fake.sentence(nb_words=3).rstrip(".")
    author = fake.name()
    publisher_name = fake.company()
    publisher_country = fake.country()
    price_base = round(random.uniform(5, 25), 2)

    book_formats = random.sample(formats, k=random.randint(1, 2))
    
    for fmt in book_formats:
        price = price_base if fmt == "Paperback" else round(price_base * random.uniform(0.6, 0.9), 2)
        flat_rows.append({
            "title": title,
            "author": author,
            "format": fmt,
            "publisher_name": publisher_name,
            "publisher_country": publisher_country,
            "price": price
        })

flat_df = pd.DataFrame(flat_rows)
flat_df
```

```python
publishers_df = flat_df[["publisher_name", "publisher_country"]].drop_duplicates().reset_index(drop=True)
publishers_df["publisher_id"] = range(1, len(publishers_df)+1)
publishers_df
```

```python
books_df = flat_df.merge(publishers_df, on=["publisher_name", "publisher_country"])
books_df = books_df[["title", "author", "format", "publisher_id", "price"]]
books_df
```

```python
rejoined_df = books_df.merge(publishers_df, on="publisher_id")
rejoined_df.equals(flat_df)
```

```python
docs = []
grouped = flat_df.groupby(["title", "author", "publisher_name", "publisher_country"])

for (title, author, pub_name, pub_country), group in grouped:
    editions = group[["format", "price"]].to_dict(orient="records")
    docs.append({
        "title": title,
        "author": author,
        "publisher": {"name": pub_name, "country": pub_country},
        "editions": editions
    })

docs[:2] 
```

```python
flat_df["price"].mean()

books_df["price"].mean()

all_prices = [edition["price"] for doc in docs for edition in doc["editions"]]
sum(all_prices)/len(all_prices)
```

```python
pub_to_update = publishers_df["publisher_name"].iloc[0]

flat_df.loc[flat_df["publisher_name"] == pub_to_update, "publisher_country"] = "New Country"

publishers_df.loc[publishers_df["publisher_name"] == pub_to_update, "publisher_country"] = "New Country"

for doc in docs:
    if doc["publisher"]["name"] == pub_to_update:
        doc["publisher"]["country"] = "New Country"
```

Finding all books by an author is easy in the flat and normalized tables because you can just filter by the author. In the document version, you have to look through each document, so it’s a bit harder.

Calculating the average price is simple in the flat and normalized tables using .mean(). In the document version, you need to go through all editions, which takes more steps.

Changing a publisher’s country is hard in the flat table because you have to update many rows. It’s very easy in the normalized table because you only update one row. In the document version, you need to update every document for that publisher.
